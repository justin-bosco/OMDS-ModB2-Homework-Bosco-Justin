# 2. Weekly graph question

Below are the graphs from last week's weekly graph question.  Suppose you wanted to cause the viewer to examine the last three months of data first.  How could you use preattentive attributes to do that?  Suggest several possibilities.

```python
import numpy as np
import seaborn as sns
import pandas as pd
```

```python
numdates = 12
np.random.seed(0)
time_series = 100 * (np.random.normal(size = numdates) / 5 + np.arange(numdates) / 16)
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Nitrate level (% of eutrophic value)")
```

```python
np.random.seed(1)
time_series = 100 * (1 - (np.random.normal(size = numdates) / 8 + np.arange(numdates) / 16))
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Phosphate level (% of eutrophic value)")
```

To make the viewer look at the last three months first, I could use these preattentive attributes:
- **Color:** draw October to December in a bright color and the rest of the line in gray.
- **Line width:** make the last three months of the line thicker than the rest.
- **Markers and labels:** add dots and data labels only on the last three months.
- **Shading:** put a light shaded box behind October to December.
- **Bold text:** add a short bold note or title that points to the last three months.

I think color with everything else in gray would work best, because the bright section stands out before the viewer reads anything else on the graph.

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 99-119 of the Storytelling With Data book as best you can. (The first half of chapter four). You do not have to get the exact data values right, just the overall look and feel.

```python
import matplotlib.pyplot as plt

segments = ['Segment 1', 'Segment 2', 'Segment 3', 'Segment 4', 'Segment 5', 'Segment 6', 'Segment 7']
us_population = [16, 7, 10, 10, 10, 32, 15]
our_customers = [9, 10, 15, 18, 17, 20, 11]
highlight = ['Segment 3', 'Segment 4', 'Segment 5']
navy = '#2b4a7a'
gray = '#bfbfbf'

fig, ax = plt.subplots(figsize=(8, 6))
for x, values in [(0, us_population), (1, our_customers)]:
    bottom = 0
    for segment, value in zip(segments, values):
        is_highlight = segment in highlight
        ax.bar(x, value, bottom=bottom, width=0.55, color=navy if is_highlight else gray, edgecolor='white')
        ax.text(x, bottom + value / 2, f'{value}%', ha='center', va='center', fontsize=12,
                color='#c9d6ea' if is_highlight else 'gray')
        bottom += value

bottom = 0
for segment, value in zip(segments, us_population):
    is_highlight = segment in highlight
    ax.text(-0.35, bottom + value / 2, segment, ha='right', va='center', fontsize=13,
            color=navy if is_highlight else 'gray', fontweight='bold' if is_highlight else 'normal')
    bottom += value

for x, values in [(0, us_population), (1, our_customers)]:
    start = sum(values[:2])
    end = sum(values[:5])
    ax.plot([x + 0.31, x + 0.31], [start, end], color=navy, linewidth=1.5)
    ax.text(x + 0.35, (start + end) / 2, f'{end - start}%', ha='left', va='center',
            fontsize=22, fontweight='bold', color=navy)

ax.set_title('Distribution by customer segment', loc='left', fontsize=17, color='#555555', x=-0.3, pad=20)
ax.set_xticks([0, 1])
ax.set_xticklabels(['US Population', 'Our Customers'], fontsize=13, color='#404040')
ax.set_xlim(-0.4, 1.6)
ax.set_ylim(0, 101)
ax.set_yticks([])
ax.tick_params(axis='x', length=0)
ax.spines[['top', 'right', 'left']].set_visible(False)
ax.spines['bottom'].set_color('lightgray')
fig.subplots_adjust(left=0.2)
plt.show()
```
